## 2. Python Lists vs NumPy Arrays

Python lists can hold mixed data types (**heterogeneous**). NumPy arrays are **homogeneous** — all elements must be the same data type. If you mix types, NumPy will upcast them to a common type.

```python
# A simple list of 10 numbers
create_list = list(range(10))
print("Items in the list:", create_list)
```

```python
def print_dtypes(l):
    for item in l:
        print(type(item))

print_dtypes(create_list)
```

```python
# A heterogeneous list (mixed data types) - perfectly fine for a Python list
list_heterogeneous = [1, "2", "A", 3, {"A": 3, "B": 4}, 9, (10, 20, 30, 40), True]
print("Heterogeneous List:\n", list_heterogeneous)
print_dtypes(list_heterogeneous)
```

```python
# NumPy array with same datatypes
a = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(a)
print("Type of a[0][0]:", type(a[0][0]))
```

```python
# Try creating a NumPy array with mixed types (int + string)
b = np.array([[1, 2, 3],
              ["A", "B", "C"]])
print(b)
print("Type of b[0][0]:", type(b[0][0]))
# Notice: NumPy silently converted the integers to strings to keep the array homogeneous!
```

## 3. Creating and Accessing Arrays

```python
arr = np.array([
    [9.0, 8.0, 7.0],
    [6.0, 5.0, 4.0]
])

print(f"Number of dimensions = {arr.ndim}")
print(f"Shape of arr        = {arr.shape}")
print(f"Total items          = {arr.size}")
print(f"Data type            = {arr.dtype}")
arr
```

```python
# Access every element with nested loops
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        print(f"Element at {i},{j} = {arr[i][j]}")
```

```python
# Select an entire row
arr[0, :]
```

```python
# Select an entire column
arr[:, 1]
```

## 5. Random Numbers & `np.random.seed()`

`np.random.seed()` makes random number generation **reproducible** — using the same seed always produces the same sequence of "random" numbers, which is important for consistent experiments.

```python
# 5x5 array from a normal distribution (mean=0, std=1), with a fixed seed
np.random.seed(10)
x = np.random.normal(0, 1, (5, 5))
x
```

```python
# Without setting the seed again, results differ every time this cell runs
x = np.random.normal(0, 1, (5, 5))
x
```

```python
# Using a different seed value produces a different (but reproducible) sequence
np.random.seed(20)
x = np.random.normal(0, 1, (5, 5))
x
```

```python
# Reset back to seed=10 -> get the EXACT same numbers as our first seed(10) run
np.random.seed(10)
x = np.random.normal(0, 1, (5, 5))
x
```

```python
print("Shape of x is:", x.shape)
```

## 6. Array Indexing

Indexing in NumPy (like Python) starts from **0**. Negative indices count from the end.

```python
x = np.array(["A", "B", "C", "D", "E", "F", "G"])
print("First element:", x[0])
print("Second element:", x[1])
print("Last element:", x[-1])
print("Second-to-last element:", x[-2])
```

```python
np.random.seed(10)
y = np.random.normal(0, 1, (2, 2))
y
```

```python
print("First row, first column:", y[0, 0])
print("Last row, last column:", y[-1, -1])
```

## 7. Array Slicing

Use slicing to access multiple elements at once: `array[start:stop:step]`.

```python
# seeting the array x to be from 10 to 29
x = np.arange(10, 30)
x
```

```python
# Elements up to (not including) index 5
x[:5]
```

```python
# Elements from index 10 onward
x[10:]
```

```python
# Elements from index 14 up to (not including) index 19
x[14:19]
```

```python
x[14:25:2]
```

```python
# Reverse the array
x[::-1]
```

## 8. Mathematics on Arrays

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(a)
```

```python
# Scalar multiplication (broadcasting)
a * 2
```

```python
# Elementwise cosine
np.cos(a)
```

```python
# Elementwise square root
np.sqrt(a.astype(float))
```

```python
# Minimum along each column (axis=0) and each row (axis=1)
print("Column-wise min:", np.min(a, axis=0))
print("Row-wise min:", np.min(a, axis=1))
```

### Elementwise Arithmetic Between Two Arrays

```python
p = np.array([[1, 2], [3, 4]], dtype=np.float64)
q = np.array([[5, 6], [7, 8]], dtype=np.float64)
```

```python
# Elementwise sum - both forms are equivalent
print(p + q)
print(np.add(p, q))
```

```python
# Elementwise difference
print(p - q)
print(np.subtract(p, q))
```

```python
# Elementwise product (NOT matrix multiplication)
print(p * q)
print(np.multiply(p, q))
```

```python
# Elementwise division
print(p / q)
print(np.divide(p, q))
```

## 9. Matrix Multiplication & Linear Algebra

```python
m1 = np.ones((2, 3))
print("m1: \n", m1)

m2 = np.full((3, 2), 2.5)
print("m2: \n",m2)

mult = np.matmul(m1, m2)
print("matrix Multiplication Result: \n" ,mult)
```

```python
# Determinant of the resulting matrix
np.linalg.det(mult)
```

### Vectors & the `.dot()` Method

```python
x2 = np.array([[1, 2], [3, 4]])
y2 = np.array([[5, 6], [7, 8]])

v = np.array([9, 10])
w = np.array([11, 12])
```

```python
# Inner product of vectors - both forms are equivalent
print(v.dot(w))
print(np.dot(v, w))
```

```python
# Matrix / vector product
print(x2.dot(v))
print(np.dot(x2, v))
```

```python
# Matrix / matrix product
print(x2.dot(y2))
print(np.dot(x2, y2))
```

## 10. Stacking and Concatenation

```python
v1 = np.array([1, 2, 3, 4])
v2 = np.array([5, 6, 7, 8])

# Stack arrays vertically (row-wise)
np.vstack([v1, v2, v1, v2])
```

```python
h1 = np.ones((2, 4))
h2 = np.zeros((2, 2))

# Stack arrays horizontally (column-wise)
np.hstack((h1, h2))
```

### `np.concatenate` with the `axis` Parameter

```python
arr_1 = np.array([1, 2, 3])
arr_2 = np.array([4, 5, 6])
arr_3 = np.array(['A', 'B', 'C'])

np.concatenate([arr_1, arr_2, arr_3])
```

```python
# Concatenation also works on 2-D arrays
matrix = np.array([["Alpha", "Beta", "Gamma"], ["Delta", "Epsilon", "Zeta"]])
print(matrix)
np.concatenate([matrix, matrix])
```

```python
# axis=0 -> stack rows on top of each other (vertical)
concatenated_axis_0 = np.concatenate([matrix, matrix], axis=0)
print(concatenated_axis_0)
print("Shape:", concatenated_axis_0.shape)
```

```python
# axis=1 -> place columns side by side (horizontal)
concatenated_axis_1 = np.concatenate([matrix, matrix], axis=1)
print(concatenated_axis_1)
print("Shape:", concatenated_axis_1.shape)
```

### Combining Arrays of Different Dimensions

`np.concatenate` requires matching dimensions, but `np.vstack` / `np.hstack` can combine a 1-D array with a 2-D array.

```python
x3 = np.array([9, 10, 11])
matrix2 = np.array([[12, 13, 14], [15, 16, 17]])
vstack_example = np.vstack([x3, matrix2])

print(vstack_example)
print("Shape of stacked array:", vstack_example.shape)
```

```python
z = np.array([[0], [0]])
hstack_example = np.hstack([matrix2, z])

print(hstack_example)
print("Shape of stacked array:", hstack_example.shape)
```
